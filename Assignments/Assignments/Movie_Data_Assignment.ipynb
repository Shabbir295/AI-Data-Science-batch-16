### **Question 30**
Dataset ko poora clean karne ke baad safely ek nayi csv file 'cleaned_movies.csv' ke naam se save karein.

```python
data = pd.read_csv('movie.csv')
data['Released_Year'] = pd.to_numeric(data['Released_Year'],errors='coerce')
data = data.dropna(subset=['Released_Year'],how='any')
data['Certificate'] = data['Certificate'].fillna('Unknown')
data['Runtime'] = data['Runtime'].str.replace(' min','').astype(int)
med = data['Meta_score'].median()
data['Meta_score'] = data['Meta_score'].fillna(med)
med_gross = data['Gross'].median()
data['Gross'] = data['Gross'].fillna(med_gross)
data.duplicated().sum()
data.to_csv('cleaned_movies.csv',index=False)
```

### **Question 38**
