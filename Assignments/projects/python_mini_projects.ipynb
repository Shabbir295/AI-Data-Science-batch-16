# 🍔 Restaurant Billing System

**📄 Features**

- 🍽️ Menu with Food Items and Prices
- 📝 Take 3 Food Items from User
- 🔍 Validate Food Items
- ❌ Handle Invalid Food Items
- 💰 Calculate Total Bill
- 🎁 Apply Discount Based on Bill Amount
- 📊 10% Discount on Bills Above 3000
- 🎉 15% Discount on Bills Above 5000
- 🧮 Calculate Final Payable Amount
- 🧾 Display Complete Restaurant Bill
- 💵 Display Individual Item Prices
- 🙏 Display Thank You Message

```python
menu = {'burger': 500,'pizza': 1200,'fries': 300,'juice': 200 , 'smartwatch': 10000}
item1 = input("Enter Your first Food")
item2 = input("Enter Your second Food")
item3 = input("Enter Your Third Food")
if item1 not in menu or item2 not in menu or item3 not in menu:
    print("Invalid Food Item! Please Enter An Item from menu")
else:
    bill_amount = menu[item1] + menu[item2] + menu[item3]
    
    if bill_amount > 5000:
        discount = bill_amount * 0.15
    elif bill_amount > 3000:
        discount = bill_amount * 0.10
    else:
        discount = bill_amount * 0
    amount_payable = bill_amount - discount
    
    print(f""" 

    ===========RESTAURANT BILL==============  
     {item1} : {menu[item1]}
     {item2} : {menu[item2]}
     {item3} : {menu[item3]}
 
     Total : {bill_amount}
     Discount : {discount}

     TotalPayable : {amount_payable}

     Thankyou For Odering
    """)
```

# 🛒 Grocery Store Billing System
**Features:**

- 🛍️ Select 3 items
- 🔍 Check whether the item is available
- ❌ Ask again if the item is invalid
- 💰 Get the price of each item
- 🧮 Calculate the total bill
- 🏷️ Apply a discount
- 💵 Calculate the final payable amount
- 📄 Display the final grocery bill

```python
products = {'rice': 300,'sugar': 180,'milk': 250,'bread': 150,'eggs': 400}
```

```python
item1 = input("Enter Your First Item").lower()
while item1 not in products:
    print("Wrong Item Enter Again")
    item1 = input("Enter your first item again").lower()
item2 = input("Enter Your First Item").lower()
while item2 not in products:
    print("Wrong item Enter Again")
    item2 = input("Enter your second item agin").lower()
item3 = input("Enter Your First Item").lower()
while item3 not in products:
    print("Wrong Item Enter Your third item again")
    item3 = input("Enter Your Third Item Again").lower()
if item1 in products:
    price1 = products[item1]
else:
    price1 = 0
    print(f"{item1} is not available at our store")
if item2 in products:
    price2 = products[item2]
else:
    price2 = 0
    print(f"{item2} is not available at our store")
if item3 in products:
    price3 = products[item3]
else:
    price3 = 0
    print(f"{item3} is not available at our store")


bill_amount = price1 + price2 + price3

if bill_amount > 3000:
    discount = bill_amount * 0.15
elif bill_amount > 1500:
    discount = bill_amount * 0.10
else:
    discount = bill_amount * 0

amount_payable = round(bill_amount - discount, 2)

print(f"""
==========GROCERY BILL==========
{item1} : {price1}
{item2} : {price2}
{item3} : {price3}
        
Total : {bill_amount}
Discount : {discount}

TotalPayable : {amount_payable}

Thankyou For Shopping!
    """)
```

# 🎓 Student Result Management System 
**Features:**

- 👤 Enter student name
- 📚 Enter marks for 5 subjects
- ❌ Ask again if marks are invalid
- 🧮 Calculate obtained marks
- 📊 Calculate percentage
- 🏆 Assign grade according to percentage
- ✅ Check Pass/Fail status
- 📄 Display the final student result

```python
name = input("Enter Your Name")
english_marks = int(input("Enter Your English Marks"))
while english_marks <0 or english_marks > 100:
    print("Invalid Marks Enter English Marks Again")
    english_marks = int(input("Enter Your English Marks again"))
math_marks = int(input("Enter Your Math Marks"))
while math_marks < 0 or math_marks > 100:
    print("Invalid Marks Enter Your Math Marks Again")
    math_marks = int(input("Enter Your Math Marks"))
physics_marks = int(input("Enter Your Physics Marks"))
while physics_marks < 0 or physics_marks > 100:
    print("Invalid Physics Marks Enter Marks Again")
    physics_marks = int(input("Enter Your Physics Marks"))
chemistry_marks = int(input("Enter Your Chemistry Marks"))
while chemistry_marks < 0 or chemistry_marks > 100:
    print("Invalid Chemistry Marks Enter Again")
    chemistry_marks = int(input("Enter Your Chemistry Marks Again"))
computer_marks = int(input("Enter Your Computer Marks"))
while computer_marks < 0 or computer_marks > 100:
    print("Invalid Computer Marks Enter Your Marks Again")
    computer_marks = int(input("Enter Your Computer Marks"))
total_marks = 500
obtained_marks = english_marks + math_marks + physics_marks + chemistry_marks + computer_marks  
percentage = obtained_marks / total_marks * 100
if percentage >= 90 and percentage <= 100:
    grade = "A+"
elif percentage >= 80 and percentage < 90 :
    grade = "A"
elif percentage >= 70 and percentage < 80:
    grade = "B"
elif percentage >= 60 and percentage < 70:
    grade = "C"
elif percentage >= 50 and percentage < 60:
    grade = "D"
elif percentage >= 0 and percentage < 50:
    grade = "F"
else:
    grade = "Invalid Marks"

if percentage >= 50:
    result = "Pass"
elif percentage >= 0 and percentage < 50:
    result = "Fail"
else:
    result = "Invalid"

print(f"""
 ========== STUDENT RESULT ==========
|                                      
| Student Name : {name}                
|                                      
| English   :      {english_marks}     
| Math      :      {math_marks}        
| Physics   :      {physics_marks}     
| Chemistry :      {chemistry_marks}    
| Computer  :      {computer_marks}    
|                                      
| Total   : {obtained_marks} / 500     
| Percentage : {percentage}%             
| Grade : {grade}                      
| Result : {result}                     
 ====================================
 """)
```

# 🏧 ATM Management System

**Features :**
- 🔐 Enter PIN
- ❌ Ask again if the PIN is incorrect
- 💰 Check account balance
- 💵 Withdraw money
- 💳 Deposit money
- ⚠️ Handle invalid amounts
- 🔄 Allow multiple transactions
- 🚪 Exit the ATM
- 📄 Display transaction details

```python
correct_pin = 1598
balance = 50000
```

```python
transaction_type = "No Transaction"
transaction_amount = 0
pin = int(input("Enter Your Pin"))
while pin != correct_pin:
    print("Wrong Pin Enter Again")
    pin = int(input("Enter Your Pin Again"))
choice = 0
while choice != 4:
    print("======== ATM MENU ========")
    print("1. Check Balance")
    print("2. Withdrawl")
    print("3. Deposit")
    print("4. Exit")
    print("5. Transaction Details")
    choice_input = int(input("Enter Your Choice"))
    choice = choice_input
    if choice == 1:
        print(balance)
    elif choice == 2:
        withdrawl_amount = int(input("Enter Your Withdrawl Amount"))
        if withdrawl_amount > 0 and withdrawl_amount <= balance:
            print("Withdrawl Allowed")
            remaining_balance = balance - withdrawl_amount
            balance = remaining_balance
            print(remaining_balance)

            transaction_type = "Withdrawal"
            transaction_amount = withdrawl_amount
        else:
            print("Insufficient Balance")
    elif choice == 3:
        deposit_amount = int(input("Enter Your Deposit Amount"))
        dp_amount = deposit_amount
        if dp_amount >= 100 and dp_amount <= 100000:
            print("Deposit Allowed")
            new_balance = dp_amount + balance
            balance  = new_balance
            print(new_balance)

            transaction_type = "Deposit"
            transaction_amount = deposit_amount
        else:
            print("Invalid Deposit Amount")
    elif choice == 4:
        print("Thank You For Using ATM") 
    elif choice == 5:
        print(f"""
        ========== TRANSACTION DETAILS ==========

        Transaction Type : {transaction_type}
        Transaction Amount : {transaction_amount}
        Current Balance : {balance}
        """)

    else:
        print("Invalid Choice! Please Select 1-5")
```
